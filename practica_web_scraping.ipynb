```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

# 1. Hacemos la peticion a la pagina web (como si fueramos el navegador)
url = "http://quotes.toscrape.com/"
respuesta = requests.get(url)

# 2. Convertimos el texto crudo del HTML en una "Sopa" que Python puede analizar
sopa = BeautifulSoup(respuesta.text, "html.parser")

# Pro-Tip: Si inspeccionas la pagina web real, veras que todas las frases estan 
# dentro de una etiqueta <span> que tiene la clase "text".
# Y todos los autores estan en una etiqueta <small> con la clase "author".

print("¡Código HTML descargado y parseado exitosamente!\n")
```

```python
frases_html = sopa.find_all('span',class_='text')
```

```python
autores_html = sopa.find_all('small', class_='author')
```

```python
primera_frase = frases_html[0].text
primer_autor = autores_html[0].text

print(primera_frase)
print(primer_autor)
```

```python
lista_frases_limpias = [frase.text for frase in frases_html]
lista_autores_limpias = [autor.text for autor in autores_html]

diccionario = {
    'Frase': lista_frases_limpias,
    'Autor': lista_autores_limpias
}

df_quotes = pd.DataFrame(diccionario)

df_quotes
```

```python
df_quotes.to_excel('frases_celebres.xlsx', index=False)

print("¡Archivo exportado exitosamente a tu computadora!")
```
