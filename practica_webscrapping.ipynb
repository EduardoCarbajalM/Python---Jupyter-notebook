```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time

url_base = "http://quotes.toscrape.com"
ruta_siguiente = "/page/1/" # Empezamos en la primera pagina

# 1. Listas maestras para guardar los datos de TODAS las paginas
todas_las_frases = []
todos_los_autores = []
pagina_actual = 1

# 2. El ciclo magico: "Mientras haya una ruta_siguiente, sigue ejecutando"
while ruta_siguiente:
    print(f"Scrapeando página {pagina_actual}...")
    url_completa = url_base + ruta_siguiente
    
    respuesta = requests.get(url_completa)
    sopa = BeautifulSoup(respuesta.text, "html.parser")
    
    # 3. Extraemos los elementos de ESTA pagina
    frases_html = sopa.find_all("span", class_="text")
    autores_html = sopa.find_all("small", class_="author")
    
    # 4. Los agregamos a nuestras listas maestras usando .extend() o un ciclo for
    for frase in frases_html:
        todas_las_frases.append(frase.text)
    for autor in autores_html:
        todos_los_autores.append(autor.text)
        
    # 5. Buscar el boton "Siguiente" en el HTML
    # Inspeccionando la web, sabemos que esta en un <li class="next">
    boton_siguiente = sopa.find("li", class_="next")
    
    if boton_siguiente:
        # Si existe el boton, extraemos el enlace de su etiqueta <a> interna
        etiqueta_a = boton_siguiente.find("a")
        ruta_siguiente = etiqueta_a["href"] # Nos dara algo como "/page/2/"
        pagina_actual += 1
        time.sleep(1) # Pro-Tip: Pausar 1 segundo para no saturar su servidor (Etiqueta de Scraper)
    else:
        # Si no hay boton (llegamos al final), rompemos el ciclo
        print("¡Llegamos a la última página!")
        ruta_siguiente = None 

# 6. Al terminar el ciclo, armamos nuestro DataFrame masivo
df_quotes_masivo = pd.DataFrame({
    "Frase": todas_las_frases,
    "Autor": todos_los_autores
})

print("\n¡Extracción masiva completada!")
```
