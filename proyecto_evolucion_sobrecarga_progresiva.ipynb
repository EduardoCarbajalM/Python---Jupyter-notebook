```python
import pandas as pd
import matplotlib.pyplot as plt

datos_gym = {
    "Semana": [1, 2, 3, 4, 5, 6],
    "Peso_Sentadilla_kg": [80, 85, 85, 90, 95, 100],
    "Peso_PressBanca_kg": [60, 62.5, 65, 65, 67.5, 70]
}
df_gym = pd.DataFrame(datos_gym)
plt.plot(df_gym['Semana'],df_gym['Peso_Sentadilla_kg'], color='blue',label='Sentadilla')
plt.plot(df_gym['Semana'], df_gym['Peso_PressBanca_kg'], color='orange', label='Press de Banca')
plt.legend()
plt.ylabel('Peso de ejercicios')
plt.xlabel('Semana')
plt.show()
```
