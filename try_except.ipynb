```python
extraccion_precios_crudos = ["120.50", "134.10", "N/A", "145.20", "ERROR_SRV", "150.00", ""]

precios_limpios = []

for precio in extraccion_precios_crudos:
    try:
        precio_float = float(precio)
        precios_limpios.append(precio_float)
    except ValueError:
        precio_float = None
        precios_limpios.append(precio_float)
        print(f"Fallo de calidad de datos ignorado: {precio}")
print(precios_limpios)
```
