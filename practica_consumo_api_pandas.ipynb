```python
import pandas as pd
import matplotlib.pyplot as plt

# Esto es lo que la API te devolvió (una Lista de Diccionarios)
respuesta_api_historico = [
    {"fecha": "2026-08-22", "moneda": "BTC", "precio_cierre": 63100.50},
    {"fecha": "2026-08-23", "moneda": "BTC", "precio_cierre": 64250.00},
    {"fecha": "2026-08-24", "moneda": "BTC", "precio_cierre": 63800.75},
    {"fecha": "2026-08-25", "moneda": "BTC", "precio_cierre": 64900.20},
    {"fecha": "2026-08-26", "moneda": "BTC", "precio_cierre": 65432.10}
]

df_respuesta = pd.DataFrame(respuesta_api_historico)

df_respuesta['fecha'] = pd.to_datetime(df_respuesta['fecha'])

plt.plot(df_respuesta['fecha'], df_respuesta['precio_cierre'])

plt.title('Precio de Bitcoin')

plt.xlabel('Fecha')

plt.ylabel('Precio de cierre')

plt.show()
```
