```python
import numpy as np

# Retornos diarios en porcentaje (negativos son pérdidas, positivos son ganancias)
retornos_diarios = np.array([-1.5, 2.3, 0.8, -0.2, -2.1, 4.5, 1.1])

dias_rentables = retornos_diarios[retornos_diarios > 0]

promedio_ganancias = np.mean(dias_rentables)

print(dias_rentables)
print(promedio_ganancias)


```

```python
import numpy as np

# Tiempos de respuesta de la API en milisegundos
latencias_ms = np.array([120, 85, 310, 45, 890, 110, 450, 92, 105, 520, 78])

latencia_promedio = np.mean(latencias_ms)
estado_servidor = np.where(latencias_ms <= 300, "SLA Cumplido", "Alerta de Rendimiento")

print(latencia_promedio)
print(estado_servidor)





```
