```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

datos_estudio = {
    "Horas_Estudio": [15, 20, 25, 30, 35, 40, 45, 50, 55, 60],
    "Calificacion_Final": [65, 68, 75, 72, 80, 85, 83, 90, 92, 98]
}
df_cert = pd.DataFrame(datos_estudio)
x = df_cert["Horas_Estudio"]
y = df_cert["Calificacion_Final"]

pendiente, interseccion, r_value, p_value, std_err = stats.linregress(x, y)

plt.scatter(x, y)

plt.plot(x, pendiente * x + interseccion, color = "red")

plt.title("Analisis de relacion entre horas de estudio y la calificacion final")

plt.xlabel("Horas de estudio")

plt.ylabel("Calificaciones finales")

plt.show()



```
