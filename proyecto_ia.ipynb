```python
import pandas as pd
import numpy as np

evaluaciones_raw = {
    "ID_Task": ["T1", "T2", "T3", "T4"],
    "ID_Anotador": ["A10", "A11", "A10", "A12"],
    "Calificacion_Dada": [4.5, 2.0, 5.0, 3.5],
    "Calificacion_Esperada": [4.0, 4.0, 5.0, 4.0]
}

catalogo_anotadores = {
    "ID_Anotador": ["A10", "A11", "A12"],
    "Nombre": ["Ana", "Carlos", "Elena"],
    "Nivel": ["Senior", "Junior", "Junior"]
}

df_evaluaciones = pd.DataFrame(evaluaciones_raw)

df_catalogo = pd.DataFrame(catalogo_anotadores)
```

```python
df_evaluaciones
```

```python
df_catalogo
```

```python
df_unificado = pd.merge(df_catalogo, df_evaluaciones, on = "ID_Anotador", how = "inner")

df_unificado
```

```python
df_unificado["Desviacion"] = np.abs((df_unificado["Calificacion_Dada"]) - (df_unificado["Calificacion_Esperada"]))
```

```python
df_recapacitacion = df_unificado.loc[df_unificado["Desviacion"] > 1]
df_recapacitacion
```
