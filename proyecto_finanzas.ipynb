```python
import pandas as pd
import numpy as np

datos_credito = {
    "ID_Cliente": ["C001", "C002", "C003", "C004", "C005"],
    "Ingreso_Mensual": [25000, np.nan, 45000, 18000, np.nan],
    "Monto_Solicitado": [100000, 30000, 150000, 25000, 80000],
    "Pagos_Atrasados": [0, 2, 0, 1, 3]
}
```

```python
df_datos = pd.DataFrame(datos_credito)

df_datos
```

```python
mediana_mensual = df_datos["Ingreso_Mensual"].median()

df_datos["Ingreso_Mensual"] = df_datos["Ingreso_Mensual"].fillna(mediana_mensual)

df_datos
```

```python
df_datos["Score_Riesgo"] = np.where((df_datos["Monto_Solicitado"] > 50000) & (df_datos["Pagos_Atrasados"] > 0), "Alto Riesgo", "Riesgo Aceptable")

df_datos
```

```python
nivel_riesgo = df_datos.groupby("Score_Riesgo")["Monto_Solicitado"].sum().reset_index()

nivel_riesgo
```
