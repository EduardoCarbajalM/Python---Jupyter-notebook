```python
import matplotlib.pyplot as plt
import pandas as pd
from scipy import stats

datos_marketing = {
    "Gasto_Publicidad_USD": [50, 60, 80, 100, 120, 150, 180, 200, 220, 250],
    "Clics_Obtenidos": [120, 140, 155, 190, 210, 260, 310, 330, 360, 400]
}
df_ads = pd.DataFrame(datos_marketing)

x = df_ads["Gasto_Publicidad_USD"]
y = df_ads["Clics_Obtenidos"]

pendiente, interseccion, r_value, p_value, std_err = stats.linregress(x, y)
plt.scatter(x, y)
plt.plot(x, pendiente * x + interseccion, color = "red")
plt.title("Tendencia de datos publicidad vs clics")
plt.xlabel("Gasto en publicidad (USD)")
plt.ylabel("Clics obtenidos")
plt.show()
```
