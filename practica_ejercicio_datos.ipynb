```python
import pandas as pd
import numpy as np
from pandas import json_normalize

# 1. Datos simulados (crudos y anidados)
filas = 50_000
datos_crudos = {
    "id_transaccion": range(1, filas + 1),
    "info_cliente": [{"pais": np.random.choice(["MX", "US", "CO"]), "vip": np.random.choice([True, False])} for _ in range(filas)],
    "monto_compra": np.random.uniform(50.0, 3000.0, size=filas),
    "departamento": np.random.choice(["Electronica", "Moda", "Hogar", "Deportes"], size=filas)
}
df_sucio = pd.DataFrame(datos_crudos)
```

```python
print(df_sucio.columns.tolist())
```

```python
df_sucio
```

```python
# 1. Aplanar la lista de diccionarios agregando el prefijo solicitado
df_junto = json_normalize(df_sucio['info_cliente']).add_prefix('info_cliente.')

# 2. Concatenar las nuevas columnas y eliminar la columna original anidada
df_sucio = pd.concat([df_sucio.drop(columns=['info_cliente']), df_junto], axis=1)
df_sucio
```

```python
archivo_parquet = 'transacciones_finales.parquet'
df_sucio.to_parquet(archivo_parquet, index=False, compression="snappy")
print(f"Archivo guardado y comprimido en Parquet exitosamente.")
```

```python
# Tomamos tu df_sucio anterior y lo guardamos particionado por departamento
df_sucio.to_parquet("dataset_particionado", partition_cols=["departamento"], index=False)
print("¡Dataset particionado guardado con éxito! Revisa tu explorador de archivos.")
```
