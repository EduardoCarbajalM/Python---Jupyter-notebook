```python
import pandas as pd
import numpy as np

datos_suscripciones = {
    "ID_Pago": ["PAY-01", "PAY-02", "PAY-03", "PAY-04", "PAY-05", "PAY-06"],
    "Fecha_Texto": ["2026-01-15", "2026-01-22", "2026-02-05", "2026-02-19", "2026-03-02", "2026-03-15"],
    "Monto": [1500.0, 2300.0, 800.0, 3100.0, 1200.0, 4500.0],
    "Plan": ["Básico", "Pro", "Básico", "Premium", "Básico", "Premium"]
}

df_pagos = pd.DataFrame(datos_suscripciones)

df_pagos
```

```python
df_pagos['Fecha_Texto'] = pd.to_datetime(df_pagos['Fecha_Texto'])

df_pagos
```

```python
df_pagos['Mes'] = df_pagos['Fecha_Texto'].dt.month

df_pagos
```

```python
df_pagos = df_pagos.sort_values('Fecha_Texto')


df_pagos
```

```python
df_pagos['Ingreso_Acumulado'] = df_pagos['Monto'].cumsum()

df_pagos
```

```python
venta_por_mes = df_pagos.groupby('Mes')['Monto'].sum().reset_index()

venta_por_mes
```
