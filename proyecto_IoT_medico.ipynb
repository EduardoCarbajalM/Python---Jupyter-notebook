```python
import pandas as pd
import numpy as np

lecturas_iot = {
    "ID_Paciente": ["P1", "P1", np.nan, "P2", "P2", "P2"],
    "Timestamp": ["10:00", "10:05", "10:10", "10:00", "10:05", "10:10"],
    "BPM": [75, 305, 80, 62, 45, np.nan]
}
```

```python
df_lecturas = pd.DataFrame(lecturas_iot)

df_lecturas
```

```python
df_lecturas = df_lecturas.dropna(subset=["ID_Paciente"])

df_lecturas
```

```python
# 1. Tu código excelente para meter los NaN a los outliers
df_lecturas["BPM"] = np.where((df_lecturas["BPM"] > 220) | (df_lecturas["BPM"] < 40), np.nan, df_lecturas["BPM"])

# 2. Calculamos los promedios sin aplastar la tabla usando transform
promedios_pacientes = df_lecturas.groupby("ID_Paciente")["BPM"].transform("mean")

# 3. Rellenamos los NaN de la columna BPM usando esos promedios calculados
df_lecturas["BPM"] = df_lecturas["BPM"].fillna(promedios_pacientes)
```
