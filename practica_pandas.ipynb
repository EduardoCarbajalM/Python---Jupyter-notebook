```python
import pandas as pd
import numpy as np

datos_api = {
    "Ticker": ["AAPL", "MSFT", "TSLA", "AMZN"],
    "Sector": ["Tecnología", "Tecnología", "Automotriz", "Retail"],
    "Precio": [175.50, 310.20, None, 135.00],
    "Volumen_Operaciones": [10200, 8500, 15000, np.nan]
}


df_mercado = pd.DataFrame(datos_api)

df_mercado


```

```python
df_mercado.info()
```

```python
precio_promedio = df_mercado["Precio"].mean()

df_mercado["Precio"] = df_mercado["Precio"].fillna(precio_promedio)

df_mercado["Volumen_Operaciones"] = df_mercado["Volumen_Operaciones"].fillna(0)

df_mercado
```

```python
df_mercado.info()
```

```python
df_tech = df_mercado.loc[df_mercado["Sector"] == "Tecnología"]

df_tech
```

```python
df_prueba = df_mercado.iloc[2:4]

df_prueba
```

```python
# Pruébalo después en tu VS Code, verás cómo Sector vuelve a ser una columna normal
volumen_por_sector = df_mercado.groupby("Sector")["Volumen_Operaciones"].sum().reset_index()
volumen_por_sector
```
