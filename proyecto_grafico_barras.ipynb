```python
import matplotlib.pyplot as plt

# Tus datos del proyecto (más dos categorías extra simuladas)
categorias = ["Electrónica", "Accesorios", "Software", "Servicios"]
ingresos = [3900.0, 2178.33, 1500.0, 950.0]
```

```python
plt.bar(categorias, ingresos, color=['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728'])
plt.title('Grafico de categorias en la tienda Electronica')
plt.xlabel('Categorias')
plt.ylabel('Ingreso de dinero')
plt.show()

```
