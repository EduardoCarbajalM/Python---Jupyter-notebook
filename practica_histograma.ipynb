```python
import matplotlib.pyplot as plt
import numpy as np

# Simulamos 500 tickets de compra. 
# loc=150 (promedio de $150), scale=40 (desviación), size=500 (cantidad de ventas)
tickets_compra = np.random.normal(loc=150, scale=40, size=500)

```

```python
plt.hist(tickets_compra, bins=15, edgecolor="black")

plt.title("Distribucion de Tickets de Compra")

plt.show()

```
