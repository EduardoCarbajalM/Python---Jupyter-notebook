```python
import pandas as pd
import numpy as np

# 1. Simulamos un dataset masivo localmente para pruebas
print("Generando dataset simulado...")
filas = 100_000
datos_grandes = {
    "id_transaccion": range(1, filas + 1),
    "categoria": np.random.choice(["Tecnología", "Hogar", "Ropa"], size=filas),
    "monto": np.random.uniform(100.0, 5000.0, size=filas),
    "fecha": pd.date_range(start="2025-01-01", periods=filas, freq="min")
}
df_masivo = pd.DataFrame(datos_grandes)

# 2. Guardado optimizado en formato Parquet (Requiere pyarrow o fastparquet)
# Nota: Para instalar motor, ejecuta: pip install pyarrow
archivo_parquet = "transacciones_optimizadas.parquet"
df_masivo.to_parquet(archivo_parquet, index=False, compression="snappy")
print(f"Archivo guardado y comprimido en Parquet exitosamente.")

# 3. Lectura eficiente de un archivo masivo por lotes (Chunking simulado para CSV)
# Si estuvieras leyendo un CSV gigante de 10 GB:
# chunk_size = 50000
# lista_chunks = []
# for chunk in pd.read_csv("archivo_gigante.csv", chunksize=chunk_size):
#     # Aplicas transformaciones al chunk antes de guardarlo en memoria
#     lista_chunks.append(chunk)
# df_final = pd.concat(lista_chunks, ignore_index=True)
```
