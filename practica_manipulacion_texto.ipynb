```python
import pandas as pd

datos_ecommerce_sucios = {
    "ID_Cliente": ["  cli-001", "CLI-002 ", " cli-003", "CLI-004"],
    "Producto_Comprado": ["LAPTOP_DELL_XPS", "mouse_logitech_mx", "TECLADO_MECANICO_Razer", "monitor_LG_4K"],
    "Monto_Pagado": ["$1,500.50", "$45.00", "$120.99", "$300.00"]
}
df_textos = pd.DataFrame(datos_ecommerce_sucios)

df_textos

df_textos['ID_Cliente'] = df_textos["ID_Cliente"].str.strip().str.upper()

df_textos

df_textos['Producto_Comprado'] = df_textos['Producto_Comprado'].str.replace('_', ' ').str.title()

df_textos

df_textos['Monto_Pagado'] = df_textos["Monto_Pagado"].str.replace('$','').str.replace(',', '').astype(float)

df_textos



```
