```python
import numpy as np
import matplotlib.pyplot as plt

# Simulamos 400 precios de relojes en el mercado (promedio $180, variación de $30)
precios_relojes = np.random.normal(loc=180, scale=30, size=400)

plt.hist(precios_relojes, bins=20, edgecolor='black')
plt.title('Distribución de Precios de Relojes Vintage')
plt.xlabel('Precios')
plt.ylabel('Cantidades')
plt.show()
```
